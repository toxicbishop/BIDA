# 7. Use a dataset of customer product reviews to classify the sentiment of each review as positive, negative, or neutral using a pre-trained machine learning model.

Using the `Naïve Bayes` Machine Learning Model

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
import os


def create_and_analyze_sentiment():
    data = {
        'review': [
            "This product is amazing! I absolutely love it.",
            "Fantastic quality and the shipping was very fast.",
            "It's an okay product, not great but not terrible either.",
            "The product is fine. It does the job as described.",
            "I am very disappointed with this purchase. It broke after one use.",
            "Horrible experience, the item was damaged on arrival.",
            "Completely useless, do not buy this product.",
            "I have mixed feelings about it. Some good parts, some bad.",
            "Works as expected. No complaints from my side.",
            "The best purchase I've made all year! Highly recommended.",
            "A total waste of money. I regret buying it.",
            "It's average. Nothing special to write home about.",
            "I'm so happy with this! It exceeded my expectations.",
            "Terrible customer service and a very poor quality product.",
            "It's a decent product for the price point."
        ],
        'sentiment': [
            'positive', 'positive', 'neutral', 'neutral', 'negative',
            'negative', 'negative', 'neutral', 'neutral', 'positive',
            'negative', 'neutral', 'positive', 'negative', 'neutral'
        ]
    }

    reviews_df = pd.DataFrame(data)
    excel_path = 'customer_reviews.xlsx'

    try:
        reviews_df.to_excel(excel_path, index=False, engine='openpyxl')
        print(f"Successfully created and saved the dataset to '{excel_path}'")
    except Exception as e:
        print(f" Error saving Excel file: {e}")
        return

    X = reviews_df['review']
    y = reviews_df['sentiment']

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=42, stratify=y
    )

    vectorizer = CountVectorizer()
    X_train_vec = vectorizer.fit_transform(X_train)
    X_test_vec = vectorizer.transform(X_test)

    model = MultinomialNB()
    model.fit(X_train_vec, y_train)

    y_pred = model.predict(X_test_vec)

    accuracy = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred, zero_division=0)

    print("\n--- Sentiment Analysis Model Evaluation ---")
    print(f"The model was trained on {len(X_train)} reviews and tested on {len(X_test)} reviews.")
    print(f"Accuracy Score: {accuracy:.2%}")

    print("\nClassification Report:")
    print(report)

    print("\n--- Example of a New Prediction ---")
    new_review = ["This is a truly wonderful and high-quality item."]
    new_review_vec = vectorizer.transform(new_review)
    prediction = model.predict(new_review_vec)

    print(f"Review: '{new_review[0]}'")
    print(f"Predicted Sentiment: {prediction[0]}")


if __name__ == '__main__':
    create_and_analyze_sentiment()
```
